## Sección A — Definición del problema

Ver detalle en `docs/taller1/definicion_problema.md`.

- **Problema:** Catálogo de servicios TI con diferencias en detalle, actualización y cumplimiento ANS → dificulta identificar demanda e incumplimientos.
- **Analítica:** Descriptiva (servicios, demanda, cumplimiento ANS, resultados, distribución). Pregunta SMART orienta clasificación predictiva para 2026.
- **Estado del arte:** Catálogo de servicios y ANS en entidades públicas para medir, organizar y apoyar decisiones.
- **Problema de IA:** Clasificación — Cumple ANS: `1` = Cumple, `0` = No cumple.
- **Impacto:** Continuidad operativa, disponibilidad TI, soporte estratégico.
- **Métricas:** Disponibilidad ≥ 99.5% · ANS ≥ 95% · Satisfacción ≥ 90% · Resolución ≤ 8 h

## Sección B — Pregunta SMART

Ver detalle en `docs/taller1/pregunta_smart.md`.

**Pregunta:** ¿Es posible predecir, durante el 2026, si un servicio de TI de la empresa cumplirá o incumplirá su ANS, utilizando los datos actuales en la atención prestada y características del servicio?

- **Específica:** Servicios TI, cumplimiento/incumplimiento ANS, atención prestada y características del servicio.
- **Medible:** Cumple ANS (1/0); métricas de clasificación y KPIs de ANS.
- **Accionable:** Anticipar riesgo y priorizar mejoras.
- **Realista:** Datos del catálogo y atención en consolidación (Drive Grupo 3).
- **Temporal:** Horizonte **2026**.
