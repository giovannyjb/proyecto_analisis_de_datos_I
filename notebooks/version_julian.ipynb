```python
# Actualizar el sivigila_data_dictionary con la información del Excel
for index, row in sivigila_data_dictionary.iterrows():
    variable_name = row['Variable']
    # Se corrige el KeyError inicial ('Variable' por 'Columna') y ahora el KeyError: 'Contexto'.
    # La columna 'Columna' en sivigila_excel_df contiene los nombres de las variables.
    match = sivigila_excel_df[sivigila_excel_df['Columna'] == variable_name]
    if not match.empty:
        # La columna 'Contexto' no existe en sivigila_excel_df. Se asume que la 'Descripcion' del Excel corresponde al 'Contexto'.
        sivigila_data_dictionary.loc[index, 'Contexto'] = match['Descripcion'].iloc[0]
        # La columna 'Expectativa' tampoco existe en sivigila_excel_df. Se dejará vacía por ahora.
        sivigila_data_dictionary.loc[index, 'Expectativa'] = ''

display(sivigila_data_dictionary)
```

Limpieza de Datos


```python
# 1. ELIMINACIÓN DE COLUMNAS COMPLETAMENTE VACÍAS O INÚTILES
# Columnas con 0 registros o con >99% de nulos que no aportan al análisis general
cols_a_eliminar = ["GRU_POB", "FEC_DEF", "CBMTE", "FM_FUERZA", "FM_UNIDAD", "FM_GRADO"]
df.drop(columns=cols_a_eliminar, inplace=True)
print(f"Columnas eliminadas: {cols_a_eliminar}")

# 2. IMPUTACIÓN DE VARIABLES CATEGÓRICAS Y CÓDIGOS
# 'COD_ASE' (Aseguradora/EPS): Imputamos faltantes con la moda o categoría 'SIN_ASEGURADORA'
moda_ase = df["COD_ASE"].mode()[0]
df["COD_ASE"] = df["COD_ASE"].fillna(moda_ase)
print(f"'COD_ASE': NaN reemplazados por la moda ({moda_ase})")

# 3. TRATAMIENTO DE FECHAS CONDICIONALES
# 'FEC_HOS' solo existe si PAC_HOS (Pacientes Hospitalizados) es verdaderamente una hospitalización.
# Convertimos a datetime; los NaN se mantienen de forma segura como NaT (Not a Time)
columnas_fechas = ["FEC_NOT", "FEC_CON", "INI_SIN", "FEC_HOS", "FECHA_NTO", "FEC_ARC_XL", "FEC_AJU"]

for col in columnas_fechas:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")

print("Formatos de fecha corregidos a tipo datetime.")

# 4. CORRECCIÓN DE TIPOS DE DATOS (TEXTO Y NÚMEROS)
# Reorganización de columnas numéricas que pandas cargó como texto/float innecesariamente
cols_enteros = ["OCUPACION"]
for col in cols_enteros:
    df[col] = df[col].fillna(0).astype("int64")

# 5. VERIFICACIÓN FINAL DEL ESTADO DE NULOS
nulos_restantes = df.isnull().sum()
print("\n--- Resumen de nulos restantes ---")
print(nulos_restantes[nulos_restantes > 0])
```

```python
sns.set_theme(style="whitegrid")

# 1. Filtrar Valle del Cauca y limpiar datos
df_valle = df[df["Departamento_residencia"].astype(str).str.contains("VALLE", case=False, na=False)].copy()
df_valle["EDAD"] = pd.to_numeric(df_valle["EDAD"], errors="coerce")
df_valle["sem_ges"] = pd.to_numeric(df_valle["sem_ges"], errors="coerce")
df_valle["SEMANA"] = pd.to_numeric(df_valle["SEMANA"], errors="coerce")

# Mapear hospitalización para mayor claridad
if "PAC_HOS" in df_valle.columns:
    df_valle["Hospitalizado"] = df_valle["PAC_HOS"].map({1: "Sí (Hospitalizado)", 2: "No (Ambulatorio)"}).fillna("Sin Info")

# ==========================================
# 7.2. CONTINUA VS. CONTINUA: DIAGRAMAS DE DISPERSIÓN
# ==========================================
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

# Dispersión 1: Edad vs Semanas de Gestación (en mujeres embarazadas)
df_gestantes = df_valle.dropna(subset=["EDAD", "sem_ges"])
if len(df_gestantes) > 0:
    sns.regplot(
        data=df_gestantes, x="EDAD", y="sem_ges", ax=axes[0],
        scatter_kws={"alpha": 0.5, "color": "teal"}, line_kws={"color": "darkred"}
    )
    r1 = df_gestantes["EDAD"].corr(df_gestantes["sem_ges"])
    axes[0].set_title(f"Edad vs. Semanas de Gestación\n(r = {r1:.2f})")
    axes[0].set_xlabel("Edad (Años)")
    axes[0].set_ylabel("Semanas de Gestación")

# Dispersión 2: Edad vs Semana Epidemiológica
df_edad_sem = df_valle.dropna(subset=["EDAD", "SEMANA"])
sns.regplot(
    data=df_edad_sem, x="SEMANA", y="EDAD", ax=axes[1],
    scatter_kws={"alpha": 0.3, "color": "steelblue"}, line_kws={"color": "darkred"}
)
r2 = df_edad_sem["SEMANA"].corr(df_edad_sem["EDAD"])
axes[1].set_title(f"Semana Epidemiológica vs. Edad\n(r = {r2:.2f})")
axes[1].set_xlabel("Semana Epidemiológica")
axes[1].set_ylabel("Edad (Años)")

plt.tight_layout()
plt.show()

# ==========================================
# 7.3. CATEGÓRICA VS. CONTINUA: EDAD SEGÚN CATEGORÍAS
# ==========================================

# 1. Edad según Estado de Hospitalización (Corrige FutureWarning con hue y legend=False)
if "Hospitalizado" in df_valle.columns:
    plt.figure(figsize=(8, 5))
    orden_hos = df_valle.groupby("Hospitalizado")["EDAD"].median().sort_values().index
    sns.boxplot(
        data=df_valle, x="Hospitalizado", y="EDAD",
        order=orden_hos, hue="Hospitalizado", palette="Set2", legend=False
    )
    plt.title("Distribución de la Edad según Estado de Hospitalización", fontweight="bold")
    plt.xlabel("Requiere Hospitalización")
    plt.ylabel("Edad (Años)")
    plt.show()

# 2. Edad según Condición Final del Caso
col_estado = "nom_est_f_caso" if "nom_est_f_caso" in df_valle.columns else "CLASIFICACION"
if col_estado in df_valle.columns:
    plt.figure(figsize=(9, 5))
    orden_est = df_valle.groupby(col_estado)["EDAD"].median().sort_values().index
    sns.boxplot(
        data=df_valle, x=col_estado, y="EDAD",
        order=orden_est, hue=col_estado, palette="Set3", legend=False
    )
    plt.title(f"Distribución de la Edad según {col_estado}", fontweight="bold")
    plt.xlabel("Estado / Diagnóstico")
    plt.ylabel("Edad (Años)")
    plt.show()

# 3. Edad por Municipios Principales del Valle del Cauca (Top 8)
top_muns = df_valle["Municipio_residencia"].value_counts().head(8).index
df_top_valle = df_valle[df_valle["Municipio_residencia"].isin(top_muns)].copy()

plt.figure(figsize=(13, 5))
orden_mun = df_top_valle.groupby("Municipio_residencia")["EDAD"].median().sort_values(ascending=False).index
sns.boxplot(
    data=df_top_valle, x="Municipio_residencia", y="EDAD",
    order=orden_mun, hue="Municipio_residencia", palette="viridis", legend=False
)
plt.xticks(rotation=45)
plt.title("Distribución de la Edad en los Municipios Más Afectados del Valle del Cauca", fontweight="bold")
plt.xlabel("Municipio")
plt.ylabel("Edad (Años)")
plt.tight_layout()
plt.show()

# ==========================================
# 7.4. VERIFICACIÓN DE UNA RELACIÓN ESPERADA: EVOLUCIÓN DE CASOS
# ==========================================
casos_por_sem = df_valle.groupby("SEMANA").size().reset_index(name="Total_Casos")

plt.figure(figsize=(9, 5))
sns.regplot(
    data=casos_por_sem, x="SEMANA", y="Total_Casos",
    scatter_kws={"alpha": 0.6, "color": "seagreen"},
    line_kws={"color": "darkred"}
)
r_temp = casos_por_sem["SEMANA"].corr(casos_por_sem["Total_Casos"])
plt.title(f"Curva de Brote: Semana Epidemiológica vs. Volumen de Casos (r = {r_temp:.2f})", fontweight="bold")
plt.xlabel("Semana Epidemiológica")
plt.ylabel("Número Total de Casos")
plt.tight_layout()
plt.show()
```
